```python
'''10. Write a python programme to calculate income tax based on annual income
upto 250000 notax , 250001-500000 5%, 500001-1000000 20%', above 1000000 : 30%'''
income = int(input("enter your annual income"))


if income <= 250000:
    tax = 0
elif income <= 500000:
    tax = (income - 250000) * 0.05
elif income <= 1000000:
    tax = (500000 - 250000) * 0.05 + (income - 500000) * 0.20
else:
    tax = (500000 - 250000) * 0.05 + (1000000 - 500000) * 0.20 + (income - 1000000) * 0.30

print("Income Tax: ", tax)
```
